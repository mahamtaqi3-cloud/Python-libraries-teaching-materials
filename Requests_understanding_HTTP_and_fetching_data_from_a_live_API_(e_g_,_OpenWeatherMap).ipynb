1. What is a website?

When you open a website such as a weather website, your browser is communicating with a server.

For example:

Your computer → Internet → Weather server

The server contains information such as:

Temperature
Humidity
Wind
Weather conditions

Your browser asks the server for this information, and the server sends it back.

2. What is HTTP?

HTTP stands for:

HyperText Transfer Protocol

Don't worry about the long name.

For now, think of HTTP as a language/rule system that allows your computer to communicate with web servers.

For example:
You: "Give me the weather data."

Server: "Sure, here it is."

That communication happens using HTTP.

3. What is a request?

When your computer asks a server for something, that is called a request.

For example:

Python → Request → Server

The server then sends a response:

Python ← Response ← Server

So:

REQUEST
Python asks for data
        ↓
     SERVER
        ↓
RESPONSE
Server sends data

4. What is an API?

Now we come to the important part.

API stands for:

Application Programming Interface

An API provides a way for one program to communicate with another program/service.

Imagine a restaurant:

You → Waiter → Kitchen
             ↓
You ← Waiter ← Food

The waiter is like an API.

You tell the waiter what you want.

The waiter communicates with the kitchen.

Then the waiter brings the result back to you.

Similarly:

Python → API → Weather Service
Python ← API ← Weather Data

5. What is a live API?

A live API provides information from a real, currently available service.

For example, a weather API can provide:

Temperature: 31°C
Humidity: 72%
Weather: Clear sky
Wind: 12 km/h

Instead of creating this data ourselves, Python gets it from the internet.

8. What does 200 mean?

HTTP uses status codes to tell us what happened.

Some important ones are:

| Code  | Meaning      |
| ----- | ------------ |
| `200` | Success ✅    |
| `400` | Bad request  |
| `401` | Unauthorized |
| `404` | Not found    |
| `500` | Server error |

So when you see:

<Response [200]>

it basically means:

The server received my request and successfully responded.

14. Sending information to the API

Suppose we want weather for Karachi.

We can give the API some parameters:

````python
params = {
    "q": "Karachi",
    "appid": "YOUR_API_KEY",
    "units": "metric"
}
````

Here:

"q": "Karachi"

means:

Search for Karachi.

"appid": "YOUR_API_KEY"

means:

Here's my API key.

"units": "metric"

means:

Give temperature in Celsius rather than Fahrenheit.

15. Making the weather request

We can then write:

````python
import requests

url = "https://api.openweathermap.org/data/2.5/weather"

params = {
    "q": "Karachi",
    "appid": "YOUR_API_KEY",
    "units": "metric"
}

response = requests.get(url, params=params)

print(response.status_code)
````

If everything is correct:
200

16. Getting the weather data

Now convert the response to JSON:

````python
data = response.json()

print(data)
````

You might receive something conceptually like:

````python
{
    "name": "Karachi",
    "main": {
        "temp": 30,
        "humidity": 70
    },
    "weather": [
        {
            "description": "clear sky"
        }
    ]
}
````

17. Extracting specific information

We don't usually want to print the entire response.

We want specific information.

City

print(data["name"])

Output:

Karachi


Temperature

Temperature is inside "main":
print(data["main"]["temp"])

Output: 30

Humidity

print(data["main"]["humidity"])

Output:

70

Weather condition
print(data["weather"][0]["description"])

Output:
clear sky

18. Complete example

Putting everything together:

````python
import requests

url = "https://api.openweathermap.org/data/2.5/weather"

params = {
    "q": "Karachi",
    "appid": "9177dd9be0326f01004c53b38bdb52bc",
    "units": "metric"
}

response = requests.get(url, params=params)

if response.status_code == 200:
    data = response.json()

    print("City:", data["name"])
    print("Temperature:", data["main"]["temp"], "°C")
    print("Humidity:", data["main"]["humidity"], "%")
    print("Weather:", data["weather"][0]["description"])

else:
    print("Something went wrong.")
````

````python
import requests

api_key = "9177dd9be0326f01004c53b38bdb52bc"
city = "Karachi"

url = "https://api.openweathermap.org/data/2.5/weather"

params = {
    "q": city,
    "appid": api_key,
    "units": "metric"
}

response = requests.get(url, params=params)

print("Status Code:", response.status_code)
print("Raw Response Text:", response.text)  # This will show you the exact error text from the server

# Only run this if the response is actually valid JSON
if response.status_code == 200:
    data = response.json()
    print("City:", data["name"])
    print("Temperature:", data["main"]["temp"], "°C")
    print("Humidity:", data["main"]["humidity"], "%")
````
