Python: Matplotlib & Seaborn — Turning Data into Charts and Plots

This lesson is a natural continuation after NumPy and Pandas. The goal is to learn how to turn a dataset into visualizations so you can quickly understand patterns, comparisons, distributions, and relationships.

1. Matplotlib

Matplotlib is one of Python's main libraries for creating graphs.

```python
import matplotlib.pyplot as plt
```

A simple line plot:

```python
x = [1, 2, 3, 4, 5]
y = [10, 15, 13, 20, 25]

plt.plot(x, y)
plt.xlabel("Day")
plt.ylabel("Value")
plt.title("Value Over Time")
plt.show()
```

2. Common Matplotlib plots
| Plot            | Used for                        |
| --------------- | ------------------------------- |
| `plt.plot()`    | Trends over time                |
| `plt.bar()`     | Comparing categories            |
| `plt.scatter()` | Relationships between variables |
| `plt.hist()`    | Distribution of values          |
| `plt.pie()`     | Parts of a whole                |


```python
#For example, a bar chart:
species = ["Cat", "Dog", "Rabbit", "Bird"]
counts = [25, 40, 15, 20]

plt.bar(species, counts)
plt.xlabel("Animal")
plt.ylabel("Count")
plt.title("Animal Counts")
plt.show()
```

3. Seaborn

Seaborn is built on top of Matplotlib and makes statistical visualizations easier and more attractive.

```python
import seaborn as sns
```

```python
#For example:
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Species": ["Cat", "Dog", "Cat", "Rabbit", "Dog", "Rabbit"],
    "Weight": [4.2, 8.5, 5.1, 2.3, 9.1, 2.8],
    "Age": [2, 5, 3, 1, 6, 2]
})

sns.scatterplot(data=df, x="Age", y="Weight", hue="Species")
plt.title("Animal Age vs Weight")
plt.show()
```

Here, hue="Species" separates the observations by species.

```python
#4. Important Seaborn plots
sns.histplot(data=df, x="Weight")
```

Shows the distribution of weight.

```python

sns.boxplot(data=df, x="Species", y="Weight")
```

Compares weight distributions between species.

```python

sns.scatterplot(data=df, x="Age", y="Weight")
```

Shows the relationship between age and weight.

```python

sns.heatmap(df.corr(numeric_only=True), annot=True)
```

Shows correlations between numerical variables.

5. Why visualization matters in biological data

Suppose you have a dataset containing:

Species

Age

Weight

Instead of looking at hundreds of rows, visualization can help you immediately see:

Temperature → Growth rate

Age → Weight

Species → Average weight

Distribution of growth rates

This is particularly useful before machine learning because graphs can reveal outliers, trends, correlations, and unusual patterns.

```python
#Example of Bar Chart in Matplotib

import matplotlib.pyplot as plt

subjects = ["Bioinformatics", "EB", "ML", "CV", "NLP"]
marks = [85, 90, 78, 93, 91]

plt.bar(subjects, marks)
plt.title("Student marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()
```

```python
#Seaborn Example (Histogram)
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.scatterplot(data=tips, x = "total_bill", y = "tip")
plt.title("total Bill vs Tip")
plt.show()

```

```python
#Seaborn Example (Histogram)
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.histplot(tips["total_bill"], bins=10)
plt.title("Distribution of Total Bill")
plt.show()
```
