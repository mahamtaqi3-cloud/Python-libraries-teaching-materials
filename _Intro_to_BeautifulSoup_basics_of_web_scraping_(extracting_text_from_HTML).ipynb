Python: Introduction to BeautifulSoup

BeautifulSoup is a Python library used to parse HTML and extract information from web pages. It is especially useful when you want to collect text, headings, links, tables, or other elements from HTML.

```python
#Then import the libraries:
from bs4 import BeautifulSoup
```

```python
#2. Start with simple HTML
#Instead of scraping a real website first, let's create some HTML ourselves:
html = """
<html>
    <head>
        <title>My Research Page</title>
    </head>
    <body>
        <h1>Python Web Scraping</h1>
        <p>BeautifulSoup is useful for extracting information from HTML.</p>
        <p>It can extract text, links, headings, and tables.</p>
    </body>
</html>
"""
```

```python
#Parse the HTML:
soup = BeautifulSoup(html, "html.parser")
```

```python
#3. Extract the page title
title = soup.title

print(title)
```

```python
#If you only want the text:
print(soup.title.text)
```

```python
#4. Extract a heading
heading = soup.h1

print(heading.text)
```

```python
#5. Extract paragraphs
#Use find_all():
paragraphs = soup.find_all("p")

for p in paragraphs:
    print(p.text)
```

This is one of the most important BeautifulSoup patterns:

```python
soup.find_all("p")
```

It means:

Find all <p> elements in the HTML.

```python
#6. Extract all text from the page
# You can use
text = soup.get_text()

print(text)
```

This extracts the text contained throughout the HTML.

For cleaner output:

```python
text = soup.get_text(separator=" ", strip=True)

print(text)
```

```python
#7. Extract links
#Let's add a link:
html = """
<html>
    <body>
        <h1>Research Resources</h1>

        <p>Learn Python for data analysis.</p>

        <a href="https://www.python.org">Python</a>
        <a href="https://pandas.pydata.org">Pandas</a>
    </body>
</html>
"""

soup = BeautifulSoup(html, "html.parser")
```

```python
#Find all links:
links = soup.find_all("a")

for link in links:
    print(link.text)
```

```python
#To get the actual URLs:
for link in links:
    print(link.get("href"))
```

```python
#You can get both text and URL:
for link in links:
    print("Name:", link.text)
    print("URL:", link.get("href"))
    print()
```

10. Important BeautifulSoup functions

| Function          | Purpose                           |
| ----------------- | --------------------------------- |
| `BeautifulSoup()` | Parse HTML                        |
| `find()`          | Find the first matching element   |
| `find_all()`      | Find all matching elements        |
| `.text`           | Extract text                      |
| `.get_text()`     | Extract text from an element/page |
| `.get("href")`    | Extract an attribute              |
| `.title`          | Access `<title>`                  |
| `.h1`             | Access `<h1>`                     |
| `.p`              | Access the first `<p>`            |


The basic workflow

Website

   ↓

requests.get()

   ↓

HTML

   ↓

BeautifulSoup()

   ↓

Parse HTML

   ↓

find() / find_all()

   ↓
   
Extract text/data

```python
from bs4 import BeautifulSoup

html = """
<html>
    <body>
        <h1>Research Project</h1>

        <p class="author">Maham Taqi</p>
        <p class="topic">Bioinformatics</p>
        <p class="topic">Python</p>
        <p class="topic">Web Scraping</p>
    </body>
</html>
"""

soup = BeautifulSoup(html, "html.parser")

# Find author
author = soup.find("p", class_="author")

print("Author:", author.text)

# Find topics
topics = soup.find_all("p", class_="topic")

for topic in topics:
    print("Topic:", topic.text)
```

```python
#3. Finding elements by ID
from bs4 import BeautifulSoup

html = """
<html>
    <body>

        <div id="main-content">
            <h1>Python Web Scraping</h1>
            <p>BeautifulSoup is used to parse HTML.</p>
        </div>

    </body>
</html>
"""

soup = BeautifulSoup(html, "html.parser")

content = soup.find(id="main-content")

print(content.text)
```

```python
#7. Navigating HTML
#BeautifulSoup allows you to move between elements.
#Example
from bs4 import BeautifulSoup

# Define sample HTML with an h1 tagPrevious element</p><h and surrounding elements
html_content = "<div><p>1>Main Title</h1><p>Next element</p></div>"

# Parse the HTML using BeautifulSoup
soup = BeautifulSoup(html_content, "html.parser")

# Find the h1 heading element
heading = soup.find("h1")

# Navigate around the HTML structure
print("Parent element:")
print(heading.parent)

print("\nNext sibling element:")
print(heading.next_sibling)

print("\nPrevious sibling element:")
print(heading.previous_sibling)
```

```python
#8. Working with tables

#This is especially useful for data analysis.

#Suppose a webpage contains:

from bs4 import BeautifulSoup

# Define the HTML table content
html_content = """
<table>
  <tr>
    <th>Species</th>
    <th>Count</th>
  </tr>
  <tr>
    <td>Oak</td>
    <td>25</td>
  </tr>
  <tr>
    <td>Pine</td>
    <td>18</td>
  </tr>
</table>
"""

# Parse the HTML
soup = BeautifulSoup(html_content, "html.parser")

# Find the table and all its rows
table = soup.find("table")
rows = table.find_all("tr")

# Loop through rows and extract data
for row in rows:
    # Find all table headers and table data cells in the current row
    cols = row.find_all(["th", "td"])
    cols = [ele.text.strip() for ele in cols]
    print(cols)
```

```python
#9. BeautifulSoup + Pandas

#This is a very useful combination for your Python learning.

import pandas as pd
import requests

url = "https://en.wikipedia.org/wiki/List_of_continents_by_population"

# Add a User-Agent header so the website doesn't block the request
headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML,"
        " like Gecko) Chrome/120.0.0.0 Safari/537.36"
    )
}

# Fetch the webpage content
response = requests.get(url, headers=headers)

# Read the HTML tables from the response text
tables = pd.read_html(response.text)

# Select the first table
df = tables[0]

# Print the DataFrame
print(df)
```

```python
#10. Clean extracted text
#Webpage text often contains unnecessary spaces and line breaks.
from bs4 import BeautifulSoup

# Define sample HTML with extra whitespace and line breaks
html_content = "<p>   \n   This is a paragraph with extra spaces and line breaks.   \n   </p>"

# Parse the HTML
soup = BeautifulSoup(html_content, "html.parser")

# Find the paragraph element
p = soup.find("p")

# 1. Without cleaning (keeps spaces/newlines)
print("Raw text:")
print(repr(p.text))

# 2. Clean text using strip=True
print("\nCleaned text (strip=True):")
print(p.get_text(strip=True))

# 3. Clean text using a custom separator and strip
print("\nCleaned text (with separator):")
text = p.get_text(" ", strip=True)
print(text)
```
