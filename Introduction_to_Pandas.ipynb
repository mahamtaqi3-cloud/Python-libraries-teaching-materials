🐼 Introduction to Pandas: Importing CSVs & Exploring a Dataset

Pandas is a Python library used for loading, cleaning, analyzing, and manipulating structured data. In bioinformatics and biological research, it is useful for working with datasets containing genes, species, samples, environmental variables, experimental results, etc.

2. Import a CSV file

Suppose you have a file called plant_growth_biological_practice.csv:

```python
import pandas as pd

df = pd.read_csv("plant_growth_biological_practice.csv")

df.head()
```

```python
#3. View the first rows
df.head()
```

```python
#You can specify the number:
df.head(3)
```

```python
#4. View the last rows
df.tail()
```

```python
#5. Check the dataset size
df.shape
```

```python
#6. See column names
df.columns
```

```python
#7. Check data types
df.dtypes
```

```python
#8. Get basic information
df.info()
```

This is one of the most useful commands when you first receive a dataset. It tells you:



Number of rows
Column names
Missing values
Data types
Memory usage

```python
#9. Get statistical summary
df.describe()
```

🧬 Biological Example

Imagine you have a dataset of plant growth:

```python
import pandas as pd

df = pd.read_csv("plant_growth_biological_practice.csv")

print(df.head())
print(df.shape)
print(df.columns)
print(df.info())
print(df.describe())
print(df.isnull().sum())
```

You could then investigate questions such as:

Does temperature affect plant growth?

For example:

df[["Temperature", "Growth"]]

Later, you could use Pandas together with NumPy, Matplotlib, Seaborn, and Scikit-learn to visualize the relationship and build a machine-learning model.

🔑 The basic Pandas workflow

CSV file → read_csv() → DataFrame → head() → shape → info() → describe() → check missing values → analyze

This is usually the first step in a data-analysis or machine-learning project.
