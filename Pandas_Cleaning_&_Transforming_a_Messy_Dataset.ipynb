Pandas: Cleaning & Transforming a Messy Dataset


In real  datasets, data is rarely perfectly organized. You may encounter missing values, duplicate rows, inconsistent spellings, incorrect data types, and unusual values. Pandas provides tools to clean and transform this data before analysis.

Step 1: Loading and Inspecting the Raw Dataset
First, we load the Excel file and inspect its structure, data types, and initial records to identify anomalies like missing values, inconsistent casing, duplicate rows, and outliers.

```python
import pandas as pd
import numpy as np

# Load the Excel file
df = pd.read_excel('students.csv')

print("--- Original DataFrame ---")
display(df)

# Check structure and missing data
print("\n--- Dataset Info ---")
df.info()
```

Identified Issues in the Raw Data:

Duplicates: Row 3 is an exact duplicate of Row 1 (Sara). (row Duplication)

Inconsistent Casing: Gender contains mixed cases (Male, female, MALE) and City has mixed cases (Karachi, karachi).

Missing Values (NaN): Ahmed is missing an Age value, and Ayesha is missing an Attendance value.

Outliers/Data Entry Errors: Zain has a Marks value of 105, which exceeds the maximum possible score of 100.

Step 2: Removing Duplicate Rows
Duplicate entries distort statistical analysis and machine learning models. We remove them using drop_duplicates().

```python
# Drop duplicate rows
df_cleaned = df.drop_duplicates().copy()

print(f"Rows remaining after removing duplicates: {len(df_cleaned)}")
```

Step 3: Standardizing Text Data

Inconsistent text formatting creates separate categories where there should be unity. We standardize Gender (Title Case) and City (Proper Title Case).

```python
# Standardize text columns
df_cleaned['Gender'] = df_cleaned['Gender'].str.capitalize()
df_cleaned['City'] = df_cleaned['City'].str.title()

print("--- After Standardizing Text ---")
display(df_cleaned[['Name', 'Gender', 'City']])
```

Step 4: Handling Missing Values

We address missing values using sensible imputation strategies:

Age: Imputed using the median age of the dataset to avoid skewing from potential extremes.

Attendance: Imputed using the mean attendance rate.

```python
# Calculate median age and mean attendance
median_age = df_cleaned['Age'].median()
mean_attendance = df_cleaned['Attendance'].mean()

# Fill missing values
df_cleaned['Age'] = df_cleaned['Age'].fillna(median_age)
df_cleaned['Attendance'] = df_cleaned['Attendance'].fillna(mean_attendance)

print("--- After Imputing Missing Values ---")
display(df_cleaned)
```

Step 5: Detecting and Correcting Outliers
Examining Marks, we notice a value of 105. We cap scores at 100 to reflect valid grading bounds.

```python
# Check marks before correction
print("Max marks before correction:", df_cleaned['Marks'].max())

# Cap marks at 100
df_cleaned.loc[df_cleaned['Marks'] > 100, 'Marks'] = 100

print("Max marks after correction:", df_cleaned['Marks'].max())
```

Step 6: Final Cleaned Dataset & Summary Statistics
With all transformations complete, we review the final dataset and generate descriptive statistics.

```python
print("--- Final Cleaned DataFrame ---")
display(df_cleaned)

print("\n--- Summary Statistics of Cleaned Data ---")
display(df_cleaned.describe())
```

Descriptive Statistics:
Age: Mean = 20.0 years, Std = 0.71, Min = 19.0, Max = 21.0

Marks: Mean = 84.4, Std = 12.70, Min = 67.0, Max = 100.0

Attendance: Mean = 0.895, Std = 0.036, Min = 0.85, Max = 0.95

1. Feature Engineering (Creating New Columns)
Often, you need to derive new insights or categories from existing data. For example, we can translate numeric marks into letter grades using a custom function and .apply():

```python
# Function to assign letter grades
def assign_grade(marks):
    if marks >= 90:
        return 'A'
    elif marks >= 80:
        return 'B'
    elif marks >= 70:
        return 'C'
    else:
        return 'D'

# Apply the function to create a new 'Grade' column
df_cleaned['Grade'] = df_cleaned['Marks'].apply(assign_grade)
display(df_cleaned[['Name', 'Marks', 'Grade']])
```

2. Advanced Outlier Detection (The IQR Method)
Instead of guessing threshold limits (like capping marks at 100), you can statistically detect outliers using the Interquartile Range (IQR):

```python
# Calculate Q1 (25th percentile) and Q3 (75th percentile)
Q1 = df_cleaned['Marks'].quantile(0.25)
Q3 = df_cleaned['Marks'].quantile(0.75)
IQR = Q3 - Q1

# Define lower and upper bounds for outliers
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Filter out outliers automatically
filtered_df = df_cleaned[(df_cleaned['Marks'] >= lower_bound) & (df_cleaned['Marks'] <= upper_bound)]
```

3. Data Type Optimization & Casting
Ensuring every column has the correct data type saves memory and prevents unexpected calculation errors:

```python
# Check current data types
print(df_cleaned.dtypes)

# Convert Age to integer (since missing values are now filled)
df_cleaned['Age'] = df_cleaned['Age'].astype(int)
```

4. Exporting Your Cleaned Dataset
Once your data is pristine, you can save it directly from Google Colab back to your local machine or Google Drive:

```python
# Save as a clean CSV file
df_cleaned.to_csv('cleaned_students.csv', index=False)

# Or save as an Excel file
df_cleaned.to_excel('cleaned_students.xlsx', index=False)
```
